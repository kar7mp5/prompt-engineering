### 4. CoT(Chain of Thought); 사고유도

질문:
```
A train travels 60km in 1.5 hours. What is the average speed?

Let's think step by step.
```

예상 답변:

```
The train travels 60km in 1.5 hours.
To calculate average speed, we divide distance by time.
60 / 1.5 = 40.
So, the average speed is 40 km/h.
```

질문:
```
If Alice is older than Bob, and Bob is older than Charlie, who is the oldest?

Let's think step by step.
```

예상 답변:
```
Alice > Bob, and Bob > Charlie.
Therefore, Alice is older than both Bob and Charlie.
Answer: Alice.
```

### 5. Reasoning; 논리유도

질문:
```
John is either in Paris or London. If he's not in Paris, where is he?

Please reason through your answer.
```

예상 답변:
```
John can only be in Paris or London.
If he's not in Paris, then he must be in London.
Therefore, the answer is London.
```

질문:
```
A patient has a fever, sore throat, and swollen lymph nodes. Based on these symptoms, what could be the cause?

Please explain your reasoning.
```

예상 답변:
```
These symptoms are commonly associated with a viral infection, particularly infectious mononucleosis.
Swollen lymph nodes and sore throat also appear in strep throat, but fever is more severe in mono.
Based on that, the likely cause is infectious mononucleosis.
```

### 6. JSON + Reason

````python
from openai import OpenAI
import os
import json

client = OpenAI(api_key=os.environ["OPENAI_API_KEY"])

response = client.responses.create(
    model="gpt-4o",
    input=[
        {
            "role": "user",
            "content": """
A user says: 'I can't log into my account and it keeps showing a 403 error.'

Classify this as a technical issue or a user mistake, and explain your reasoning.

Respond in JSON:
{
  "category": "...",
  "reason": "..."
}
"""
        }
    ]
)

# 출력 및 JSON 파싱 시도
raw_output = response.output_text
print("Raw output:\n", raw_output)

try:
    result = json.loads(raw_output)
    print("\nParsed JSON:")
    print("Category:", result["category"])
    print("Reason:", result["reason"])
except json.JSONDecodeError:
    print("\nJSON parsing failed. GPT might have output extra text.")
````
