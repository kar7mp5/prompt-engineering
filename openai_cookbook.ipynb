### 1. 단순 요약 프롬프트

```
다음 기사 내용을 두 문장으로 요약해줘:

"2024년 총선 결과, 여당은 전체 의석의 60%를 차지하며 압승..."
```

```
다음 문장에서 인물, 날짜, 장소를 JSON으로 추출해줘:

"2023년 11월 5일, 엘론 머스크는 도쿄에서 AI 회의를 개최했다."
```

```json
{
  "person": "엘론 머스크",
  "date": "2023년 11월 5일",
  "location": "도쿄"
}
```

```
너는 마케팅 전문가야.

각 문구는 1문장 이내여야 해.

제품 설명: 스마트폰 거치대, 접이식, 차량용, 자석 부착 가능
```
